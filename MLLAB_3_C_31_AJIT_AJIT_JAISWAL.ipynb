### 1. Import libraries and load dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv("winequality-red.csv")

df.head(10)
df.tail(10)

df.info()
df.shape
df.describe()
```

### 2. Find-S Algorithm

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

def find_s(X, y):
    hypo = None
    print("Initial Hypothesis:", hypo)

    for i in range(len(X)):
        if y[i] == "acceptable":
            if hypo is None:
                hypo = X[i].copy()
            else:
                for j in range(len(hypo)):
                    if hypo[j] != X[i][j]:
                        hypo[j] = '?'

        print(f"After Training Example {i+1}")
        print(hypo)

    return hypo

final_hypo = find_s(X, y)
print("\nFinal Hypothesis")
print(final_hypo)
```

### 3. Candidate Elimination Algorithm

```python
import copy

def candidate_elimination(concepts, target):
    specific = [concepts[0].copy()]
    general = [["?" for _ in range(len(specific[0]))]]

    print("Initial S:", specific)
    print("Initial G:", general)

    for i, h in enumerate(concepts):
        if target[i] == "acceptable":

            for x in range(len(specific)):
                if h[x] != specific[x][x]:
                    specific[x][x] = '?'

        else:

            for x in range(len(specific)):
                if h[x] == specific[x][x]:
                    general.append(
                        ["?" if j != x else specific[x][x]
                         for j in range(len(specific))]
                    )

        print("\nExample", i + 1)
        print("S =", specific)
        print("G =", general)

    return specific, general


S, G = candidate_elimination(X, y)

print("\nFinal Specific Boundary")
print(S)

print("\nFinal General Boundary")
for g in G:
    print(g)
```
